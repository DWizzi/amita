```python
import numpy as np
import polars as pl
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

data generating process:

$Y = 0.3X_1 + 0.4X_2 + \epsilon_i\\$
$X_1 \sim \mathrm{exp}(3) \\$
$X_2 \sim \mathcal{N}(2, 1) \\$
$\epsilon \sim \mathcal{N}(0, 0.5)$

```python
np.random.seed(1010)

x1 = np.random.exponential(scale=3, size=30)
x2 = np.random.normal(2, 1, 30)
e = np.random.normal(0, 0.5, 30)

y = 0.3 * x1 + 0.4 * x2 + e

X = np.concatenate(
    [
        x1.reshape((30,1)),
        x2.reshape((30,1)),
    ], 
    axis=1
)

pl.DataFrame(
    data=np.concatenate([
        y.reshape((30,1)),
        x1.reshape((30,1)),
        x2.reshape((30,1)),
    ], axis=1),
    schema=["y", "x1", "x2"]
)

print(y)
print(X)
```

```python
sm.OLS(y, X).fit().summary()
```

```python
sm.OLS(y, X).fit(kwargs={"cov_type": "HC3"}).summary()
```
